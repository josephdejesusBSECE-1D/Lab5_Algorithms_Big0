```python
# Student configuration
LAST_NAME = "De Jesus"
STUDENT_ID = "TUPM-26-1944"

SEED_DIGIT = int(STUDENT_ID[-1])
ID_SUM = sum(int(digit) for digit in STUDENT_ID if digit.isdigit())
NAME_LENGTH = len(LAST_NAME)

print("Seed:", SEED_DIGIT)
print("ID Sum:", ID_SUM)
print("Name Length:", NAME_LENGTH)
```

```python
# CELL 2 - LARGEST-VALUE ALGORITHM
data = [
    10 + SEED_DIGIT,
    20 + SEED_DIGIT,
    5 + NAME_LENGTH,
    8 + NAME_LENGTH,
    15 + ID_SUM,
]
largest = data[0]
comparisons = 0
for value in data[1:]:
    comparisons += 1
    if value > largest:
        largest = value

print("Data:", data)
print("Comparisons:", comparisons)
print("Largest:", largest)
```

```python
# CELL 3 - LINEAR GROWTH
def linear_operations(n):
    count = 0
    for _ in range(n):
        count += 1
    return count

for n in [10, 100, 1000, 5000, 10000]:
    print(f"n={n:5d} | operations={linear_operations(n):5d}")
```

```python
# CELL 4 - BIG O CLASSIFICATION
def log_ops(n):
    count = 0
    while n > 1:
        n //= 2
        count += 1
    return count

for n in [10, 100, 1000]:
    print(
        f"n={n:4d} | "
        f"O(log n)={log_ops(n):2d} | "
        f"O(n)={n:4d} | "
        f"O(n^2)={n * n:7d} | "
        "O(1)=1"
    )
```

```python
# CELL 5 - SPACE COMPLEXITY
def constant_space(n):
    total = 0
    for value in range(n):
        total += value
    return total


def linear_space(n):
    return list(range(n))

for n in [10, 100, 1000, 5000]:
    data = linear_space(n)
    print(f"n={n:5d} | O(1): constant | O(n): {len(data):5d}")
```

```python
# CELL 6 - BEST/AVERAGE/WORST CASE
def search_case(data, target):
    for comparisons, value in enumerate(data, 1):
        if value == target:
            return comparisons
    return len(data)


data = [11, 22, 33, 44, 55]
for case, target in {"Best": 11, "Average": 33, "Worst": 99}.items():
    print(f"{case} Case: {search_case(data, target)} comparisons")
```

```python
# CELL 7 - LINEAR SEARCH
def linear_search(data, target):
    for index, value in enumerate(data):
        if value == target:
            return index, index + 1
    return -1, len(data)


data = list(range(1, 21))
target = SEED_DIGIT + 1
index, comparisons = linear_search(data, target)
print("Target:", target)
print("Index:", index)
print("Comparisons:", comparisons)
```

```python
# CELL 8 - BINARY SEARCH
def binary_search(data, target):
    low, high, comparisons = 0, len(data) - 1, 0
    while low <= high:
        middle = (low + high) // 2
        comparisons += 1
        if data[middle] == target:
            return middle, comparisons
        if data[middle] < target:
            low = middle + 1
        else:
            high = middle - 1
    return -1, comparisons


index, binary_comparisons = binary_search(data, target)
print("Target:", target)
print("Index:", index)
print("Comparisons:", binary_comparisons)
```

```python
# CELL 9 - SEARCH COMPARISON
data = list(range(1, 101))
target = ((ID_SUM + SEED_DIGIT) % 100) + 1
linear_index, linear_comparisons = linear_search(data, target)
binary_index, binary_comparisons = binary_search(data, target)
print("Target:", target)
print(f"Linear -> index={linear_index}, comparisons={linear_comparisons}")
print(f"Binary -> index={binary_index}, comparisons={binary_comparisons}")
print("Reduction:", linear_comparisons - binary_comparisons)
```

```python
# CELL 10 - BUBBLE SORT
def bubble_sort(data):
    values = data.copy()
    comparisons = swaps = 0
    for end in range(len(values)):
        swapped = False
        for index in range(len(values) - end - 1):
            comparisons += 1
            if values[index] > values[index + 1]:
                values[index], values[index + 1] = values[index + 1], values[index]
                swaps += 1
                swapped = True
        if not swapped:
            break
    return values, comparisons, swaps


data = [
    (ID_SUM * 3) % 100,
    (SEED_DIGIT * 11) % 100,
    (NAME_LENGTH * 7) % 100,
    42, 17, 89, 31, 56,
]
sorted_data, bubble_comparisons, bubble_swaps = bubble_sort(data)
print("Bubble Sorted:", sorted_data)
print("Comparisons:", bubble_comparisons, "| Swaps:", bubble_swaps)
```

```python
# CELL 11 - SELECTION SORT
def selection_sort(data):
    values = data.copy()
    comparisons = swaps = 0
    for start in range(len(values) - 1):
        minimum = start
        for index in range(start + 1, len(values)):
            comparisons += 1
            if values[index] < values[minimum]:
                minimum = index
        if minimum != start:
            values[start], values[minimum] = values[minimum], values[start]
            swaps += 1
    return values, comparisons, swaps


sorted_data, selection_comparisons, selection_swaps = selection_sort(data)
print("Selection Sorted:", sorted_data)
print("Comparisons:", selection_comparisons, "| Swaps:", selection_swaps)
```

```python
# CELL 12 - INSERTION SORT
def insertion_sort(data):
    values = data.copy()
    comparisons = shifts = 0
    for index in range(1, len(values)):
        key = values[index]
        previous = index - 1
        while previous >= 0:
            comparisons += 1
            if values[previous] > key:
                values[previous + 1] = values[previous]
                shifts += 1
                previous -= 1
            else:
                break
        values[previous + 1] = key
    return values, comparisons, shifts


sorted_data, insertion_comparisons, insertion_shifts = insertion_sort(data)
print("Insertion Sorted:", sorted_data)
print("Comparisons:", insertion_comparisons, "| Shifts:", insertion_shifts)
```

```python
# CELL 13 - EXECUTION-TIME MEASUREMENT
import time


def measure(sort_function, data, trials=5):
    times = []
    result = None
    for _ in range(trials):
        start = time.perf_counter()
        result = sort_function(data)
        times.append(time.perf_counter() - start)
    return result, sum(times) / trials


data_size = 100 + ID_SUM * 10
performance_data = [(ID_SUM * 3 + index * 7) % 1000 for index in range(data_size)]

bubble_result, bubble_time = measure(bubble_sort, performance_data)
selection_result, selection_time = measure(selection_sort, performance_data)
insertion_result, insertion_time = measure(insertion_sort, performance_data)

print(f"Dataset Size: {data_size}")
print(f"Bubble Sort:    {bubble_time:.8f}s")
print(f"Selection Sort: {selection_time:.8f}s")
print(f"Insertion Sort: {insertion_time:.8f}s")
```

```python
# CELL 14 - SORTING PERFORMANCE COMPARISON
results = [
    ("Bubble Sort", bubble_result),
    ("Selection Sort", selection_result),
    ("Insertion Sort", insertion_result),
]

print(f"{'Algorithm':<18} {'Comparisons':>12} {'Movements':>12} {'Time (s)':>14}")
for (name, result), elapsed in zip(results, [bubble_time, selection_time, insertion_time]):
    print(f"{name:<18} {result[1]:>12} {result[2]:>12} {elapsed:>14.8f}")
```
