```python
LAST_NAME = "De Jesus"
STUDENT_ID = "TUPM-26-1944"
FAVORITE_ARTIST = "Amiel Sol"

SEED_DIGIT = int(STUDENT_ID[-1])
ID_SUM = sum(int(digit) for digit in STUDENT_ID if digit.isdigit())
ARTIST_SUM = sum(ord(letter.upper()) - 64 for letter in FAVORITE_ARTIST if letter.isalpha())

start = ID_SUM + ARTIST_SUM
original = [start + index * SEED_DIGIT for index in range(10)]
rotation = SEED_DIGIT % len(original)
rotated = original[rotation:] + original[:rotation]
target_index = (ID_SUM + ARTIST_SUM) % len(original)
target = original[target_index]

def rotated_search(data, target):
    low, high = 0, len(data) - 1
    comparisons = 0

    while low <= high:
        middle = (low + high) // 2
        comparisons += 1

        if data[middle] == target:
            return middle, comparisons

        if data[low] <= data[middle]:
            if data[low] <= target < data[middle]:
                high = middle - 1
            else:
                low = middle + 1
        else:
            if data[middle] < target <= data[high]:
                low = middle + 1
            else:
                high = middle - 1

    return -1, comparisons

def linear_search(data, target):
    comparisons = 0
    for index, value in enumerate(data):
        comparisons += 1
        if value == target:
            return index, comparisons
    return -1, comparisons

position, comparisons = rotated_search(rotated, target)
linear_position, linear_comparisons = linear_search(rotated, target)
expected_position = (target_index - rotation) % len(original)

assert original == sorted(original)
assert rotated == original[rotation:] + original[:rotation]
assert position == expected_position and linear_position == position

print("Original Dataset:", original)
print("Rotated Dataset:", rotated)
print("Target Incident:", target)
print("Target Position (zero-based):", position)
print("Binary Search Comparisons:", comparisons)
print("Linear Search Comparisons:", linear_comparisons)
print("Time Complexity: O(log n)")
print("Space Complexity: O(1)")
print(f"Final Output: Target Incident found at position {position}")
```

```python
NAME_LENGTH = len(LAST_NAME)
ARTIST_LENGTH = len(FAVORITE_ARTIST)

dataset_A = [ID_SUM * SEED_DIGIT * index for index in range(5)]
dataset_B = [ARTIST_LENGTH + NAME_LENGTH * index for index in range(5)]

def merge(first, second):
    result = []
    first_index = second_index = comparisons = 0

    while first_index < len(first) and second_index < len(second):
        comparisons += 1
        if first[first_index] <= second[second_index]:
            result.append(first[first_index])
            first_index += 1
        else:
            result.append(second[second_index])
            second_index += 1

    result.extend(first[first_index:])
    result.extend(second[second_index:])
    return result, comparisons

combined, merge_comparisons = merge(dataset_A, dataset_B)
assert combined == sorted(dataset_A + dataset_B)
assert len(combined) == len(dataset_A) + len(dataset_B)

print("Department A dataset:", dataset_A)
print("Department B dataset:", dataset_B)
print("Consolidated dataset:", combined)
print("Merge comparisons:", merge_comparisons)
print("Number of Records Before Consolidation:", len(dataset_A) + len(dataset_B))
print("Number of Records After Consolidation:", len(combined))
print("Time Complexity: O(n + m)")
print("Space Complexity: O(n + m)")
print("Final Output: Enrollment records successfully consolidated.")
```

```python
import random
import time

orders = [ID_SUM + SEED_DIGIT * index + NAME_LENGTH for index in range(10)]
sorted_input = sorted(orders)
reverse_input = sorted(orders, reverse=True)
random_input = orders.copy()
random.seed(SEED_DIGIT)
random.shuffle(random_input)

def bubble_sort(data):
    values = data.copy()
    comparisons = movements = 0
    start_time = time.perf_counter()
    for end in range(len(values)):
        for index in range(len(values) - 1 - end):
            comparisons += 1
            if values[index] > values[index + 1]:
                values[index], values[index + 1] = values[index + 1], values[index]
                movements += 1
    return values, comparisons, movements, time.perf_counter() - start_time

def selection_sort(data):
    values = data.copy()
    comparisons = movements = 0
    start_time = time.perf_counter()
    for start in range(len(values) - 1):
        minimum = start
        for index in range(start + 1, len(values)):
            comparisons += 1
            if values[index] < values[minimum]:
                minimum = index
        if minimum != start:
            values[start], values[minimum] = values[minimum], values[start]
            movements += 1
    return values, comparisons, movements, time.perf_counter() - start_time

def insertion_sort(data):
    values = data.copy()
    comparisons = movements = 0
    start_time = time.perf_counter()
    for index in range(1, len(values)):
        key = values[index]
        previous = index - 1
        while previous >= 0:
            comparisons += 1
            if values[previous] > key:
                values[previous + 1] = values[previous]
                movements += 1
                previous -= 1
            else:
                break
        values[previous + 1] = key
    return values, comparisons, movements, time.perf_counter() - start_time

input_sets = {"Sorted": sorted_input, "Reverse": reverse_input, "Random": random_input}
algorithms = {"Bubble Sort": bubble_sort, "Selection Sort": selection_sort, "Insertion Sort": insertion_sort}

print("Generated Order Dataset:", orders)
print("Sorted Input:", sorted_input)
print("Reverse Input:", reverse_input)
print("Random Input:", random_input)

for condition, data in input_sets.items():
    print(f"\nInput Condition: {condition}")
    for name, algorithm in algorithms.items():
        result, comparisons, movements, elapsed = algorithm(data)
        assert result == sorted(data), f"{name} failed for {condition} input"
        print(f"\n{name}")
        print("Sorted Result:", result)
        print("Comparisons:", comparisons)
        print("Movements:", movements)
        print(f"Execution time: {elapsed:.8f} seconds")
        print("Sorting Verification: True")

print("\n--- BIG O ANALYSIS ---")
print("Bubble Sort: O(n^2)")
print("Selection Sort: O(n^2)")
print("Insertion Sort: Best O(n), Average/Worst O(n^2)")
print("\nBest Case: Sorted Input")
print("Average Case: Random Input")
print("Worst Case: Reverse Input")
print("\nFinal Output: All sorting algorithms completed successfully.")
```
